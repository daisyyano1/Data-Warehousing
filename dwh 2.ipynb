```python
import sqlite3
import pandas as pd

# connect to the database
conn = sqlite3.connect("data_warehouse.db")

# create the employees table if it does not exist
conn.execute("""
CREATE TABLE IF NOT EXISTS employees (
	id INTEGER PRIMARY KEY,
	name TEXT,
	age INTEGER,
	salary INTEGER
)
""")

# insert sample data if table is empty
if pd.read_sql("SELECT COUNT(*) as cnt FROM employees;", conn)['cnt'][0] == 0:
	sample_data = [
		(1, 'Alice', 24, 50000),
		(2, 'Bob', 30, 60000),
		(3, 'Charlie', 22, 55000),
		(4, 'David', 35, 70000),
		(5, 'Eva', 28, 65000)
	]
	conn.executemany("INSERT INTO employees (id, name, age, salary) VALUES (?, ?, ?, ?);", sample_data)
	conn.commit()

# list all tables
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("Tables in database:")
display(tables)

# fetch data from the 'employees' table and display as a DataFrame
df_employees = pd.read_sql("SELECT * FROM employees;", conn)
print("Employee Table Data:")
display(df_employees)

conn.close()

```
